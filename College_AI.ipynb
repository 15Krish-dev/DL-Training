```python
from langchain_community.document_loaders import PyPDFLoader

files = [
    "Student Code of Conduct and IT Guidelines.pdf",
    "Scholarships and Fee Refund Guidelines.pdf",
    "Campus Housing and Residential Life Regulations.pdf",
    "College Academic Regulations & Grading Policy.pdf"
]

text_documents = []
for file in files:
    loader = PyPDFLoader(file)
    text_documents.extend(loader.load())

print(f"Total pages loaded: {len(text_documents)}")
```

```python
# Text splitter
from langchain_text_splitters import RecursiveCharacterTextSplitter
textSplitter = RecursiveCharacterTextSplitter(chunk_size=200,chunk_overlap=20)
finalDocuments = textSplitter.split_documents(text_documents)
finalDocuments
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
# Embedding and vector store in vectordatabase
from langchain_community.vectorstores import FAISS
db = FAISS.from_documents(finalDocuments,embeddings)
db
```

```python
query = input("Enter your question for C++: ")
docs = db.similarity_search(query)
for i in range(len(docs)):
    print(docs[i].page_content)
```
